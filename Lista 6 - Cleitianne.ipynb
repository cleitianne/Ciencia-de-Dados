### c) Agrupe as linhas com postagens iguais ou extremamente semelhantes. Aqui você pode utilizar uma métrica de semelhança de textos. Crie uma variável para representar a quantidade de vezes que a mensagem foi compartilhada. Observe que ao agrupar linhas que possuem a “mesma” postagem (texto), você deve escolher como valor para as variáveis data e hora da postagem, os valores da cópia mais antiga. 


```python
# Importando as bibliotecas necessárias para a tarefa
from thefuzz import process, fuzz
import pandas as pd

# --- Etapa 1: Preparação Adicional dos Dados ---
# O seu código já criou 'df_limpo'. Vamos garantir que os tipos de dados estão corretos para a operação.

# Garantir que a coluna de data esteja no formato datetime para podermos encontrar a mais antiga
df_limpo['date_message'] = pd.to_datetime(df_limpo['date_message'])

# Preencher possíveis valores nulos (NaN) na coluna de texto para evitar erros no processamento
df_limpo['text_content_anonymous'] = df_limpo['text_content_anonymous'].fillna('')

# Criar uma cópia de trabalho para evitar o SettingWithCopyWarning do pandas
df_to_process = df_limpo.copy()


# --- Etapa 2: Lógica de Agrupamento por Similaridade (VERSÃO CORRIGIDA) ---

# Limiar de similaridade. Textos com 90% ou mais de semelhança serão agrupados.
SIMILARITY_THRESHOLD = 90

# Armazena os clusters (grupos) de mensagens similares.
clusters = []
# Guarda os índices que já foram agrupados para não os processarmos novamente.
processed_indices = set()

# Criar um dicionário (índice -> texto) é mais eficiente para a busca.
choices_dict = df_to_process['text_content_anonymous'].to_dict()

print(f"Iniciando o agrupamento por similaridade para {len(df_to_process)} mensagens...")

for index, row in df_to_process.iterrows():
    # Pula a mensagem se ela já foi adicionada a um grupo
    if index in processed_indices:
        continue

    # Usa o texto da linha atual como consulta para encontrar todos os textos similares no dataset
    query_text = row['text_content_anonymous']
    
    # CORREÇÃO: Usar 'process.extractBests' que aceita 'score_cutoff'.
    # Adicionamos 'limit=None' para obter todos os resultados que atendem ao critério.
    similar_texts = process.extractBests(
        query_text, 
        choices_dict, 
        scorer=fuzz.ratio, 
        score_cutoff=SIMILARITY_THRESHOLD,
        limit=None
    )

    # Extrai apenas os índices do DataFrame das mensagens que formam o cluster
    current_cluster_indices = [idx for text, score, idx in similar_texts]
    
    # Adiciona o novo grupo à nossa lista de clusters
    if current_cluster_indices:
        clusters.append(current_cluster_indices)

        # Marca todos os índices deste novo grupo como processados para não os checar novamente
        processed_indices.update(current_cluster_indices)

print(f"Agrupamento concluído. Encontrados {len(clusters)} grupos de mensagens únicas/similares.")

# --- Etapa 3: Processando os Grupos e Criando o DataFrame Final ---

final_rows = []

for cluster_indices in clusters:
    # Seleciona todas as linhas do DataFrame original que pertencem ao grupo atual
    cluster_df = df_to_process.loc[cluster_indices]
    
    # Encontra o índice da mensagem MAIS ANTIGA no grupo
    oldest_index = cluster_df['date_message'].idxmin()
    
    # Usa a linha da mensagem mais antiga como a representante do grupo
    representative_row = cluster_df.loc[oldest_index].copy()
    
    # Cria a nova coluna com a contagem de compartilhamentos (o tamanho do grupo)
    representative_row['share_count'] = len(cluster_df)
    
    # Adiciona a linha representativa à nossa lista de resultados
    final_rows.append(representative_row)

# Cria o DataFrame final com os resultados agrupados
df_agrupado = pd.DataFrame(final_rows)

# Ordena o resultado pela data e reseta o índice para um resultado limpo
df_agrupado = df_agrupado.sort_values('date_message').reset_index(drop=True)


# --- Etapa 4: Exibindo o Resultado ---

print("\n--- DataFrame Final Agrupado ---")
# Exibindo as colunas mais relevantes para verificação do resultado
display_cols = ['date_message', 'share_count', 'text_content_anonymous']
print(df_agrupado[display_cols].head())

# Para ver um exemplo de mensagens que foram efetivamente agrupadas
print("\nExemplo de mensagens que foram agrupadas (com contagem de compartilhamento > 1):")
print(df_agrupado[df_agrupado['share_count'] > 1][display_cols])
```

### d) Você pode criar novos atributos numéricos, tais como: quantidade de palavras, quantidade de caracteres etc. 


```python
import string # Biblioteca necessária para a contagem de pontuação

# --- Etapa 5: Criação de Novos Atributos Numéricos ---
# Usando o DataFrame 'df_agrupado' da célula anterior.

print("Criando novos atributos numéricos a partir do texto...")

# 1. Quantidade de Caracteres: O tamanho total da string.
df_agrupado['char_count'] = df_agrupado['text_content_anonymous'].str.len()

# 2. Quantidade de Palavras: Divide o texto por espaços e conta o número de elementos.
# O fillna(0) trata casos onde o texto pode ser nulo (apesar de termos tratado antes, é uma boa prática).
df_agrupado['word_count'] = df_agrupado['text_content_anonymous'].str.split().str.len().fillna(0).astype(int)

# 3. Comprimento Médio da Palavra: Pode indicar a complexidade do vocabulário.
# Dividimos o total de caracteres pelo total de palavras.
# Usamos um truque para evitar a divisão por zero para mensagens sem palavras.
df_agrupado['avg_word_length'] = df_agrupado['char_count'] / df_agrupado['word_count']
# Substitui possíveis valores infinitos ou nulos (resultantes da divisão por 0) por 0.
df_agrupado['avg_word_length'].fillna(0, inplace=True)

# 4. Contagem de Pontuação: Conta o número de caracteres de pontuação.
# Pode ser um indicador de tom ou urgência (ex: "!!!", "???").
df_agrupado['punctuation_count'] = df_agrupado['text_content_anonymous'].apply(
    lambda text: sum(1 for char in text if char in string.punctuation)
)

# 5. Proporção de Letras Maiúsculas: Indica ênfase ou "grito".
# Um valor alto é comum em correntes e mensagens de desinformação.
df_agrupado['uppercase_ratio'] = df_agrupado['text_content_anonymous'].apply(
    lambda text: sum(1 for char in text if char.isupper()) / len(text) if len(text) > 0 else 0
)


# --- Exibindo o Resultado ---
print("Novos atributos criados com sucesso!")

# Seleciona as colunas originais mais relevantes e as novas colunas para uma visualização clara
display_cols = [
    'text_content_anonymous',
    'share_count',
    'char_count',
    'word_count',
    'avg_word_length',
    'punctuation_count',
    'uppercase_ratio'
]

# Mostra as primeiras 5 linhas do DataFrame com as novas colunas
print("\n--- DataFrame com Novos Atributos ---")
print(df_agrupado[display_cols].head())

# Mostra estatísticas descritivas das novas colunas para ter uma ideia da distribuição
print("\n--- Estatísticas Descritivas dos Novos Atributos ---")
print(df_agrupado[display_cols[2:]].describe())
```

### Questão 1
### Utilizando os dados referente a postagens no Telegram, crie um modelo preditivo (classificador binário) para classificar uma mensagem em duas classes possíveis: “viral” (classe positiva) ou “não viral” (classe negativa). Para rotular as mensagens únicas (agrupadas) nas classes, “viral” e “não viral”, utilize a seguinte estratégia: Calcule um limiar (threshold). Por exemplo, mediana do número de compartilhamentos mais dois desvios padrões. As mensagens com quantidade de compartilhamentos maiores ou iguais ao limiar definido devem ser rotuladas como “virais”. As demais mensagens devem ser rotuladas como “não virais”. 

```python
import pandas as pd
import numpy as np
import string
import nltk

# Importando componentes do Scikit-learn
from sklearn.model_selection import KFold
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from scipy.sparse import hstack

# Importando componentes para pré-processamento de texto
from nltk.corpus import stopwords
from nltk.stem import RSLPStemmer

try:
    # Checa se as stopwords para português estão disponíveis
    stopwords.words('portuguese')
except LookupError:
    print("Baixando o recurso 'stopwords' do NLTK...")
    nltk.download('stopwords')

try:
    # Tenta instanciar o stemmer para forçar a verificação do recurso 'rslp'
    RSLPStemmer()
except LookupError:
    print("Baixando o recurso 'rslp' do NLTK...")
    nltk.download('rslp')

print("Recursos do NLTK estão prontos.")

# ==============================================================================
# 1. DEFINIÇÃO DA VARIÁVEL ALVO (VIRAL / NÃO VIRAL)
# ==============================================================================
print("--- 1. Definindo a variável alvo 'is_viral' ---")

share_median = df_agrupado['share_count'].median()
share_std = df_agrupado['share_count'].std()
threshold = share_median + (2 * share_std)

print(f"Mediana de compartilhamentos: {share_median:.2f}")
print(f"Desvio Padrão de compartilhamentos: {share_std:.2f}")
print(f"Limiar de Viralização (threshold) calculado: {threshold:.2f}")

df_agrupado['is_viral'] = (df_agrupado['share_count'] >= threshold).astype(int)

print("\nDistribuição das classes:")
print(df_agrupado['is_viral'].value_counts())

if len(df_agrupado['is_viral'].unique()) < 2:
    raise ValueError("O limiar resultou em apenas uma classe. Não é possível treinar um classificador.")

print("\nVariável alvo criada com sucesso.")


# ==============================================================================
# 2. FUNÇÕES AUXILIARES E PRÉ-PROCESSAMENTO (Item h)
# ==============================================================================
print("\n--- 2. Definindo funções de pré-processamento de texto ---")

stop_words_pt = set(stopwords.words('portuguese'))
stemmer_pt = RSLPStemmer()

def preprocess_text(text):
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = ''.join([char for char in text if char not in string.punctuation])
    tokens = [word for word in text.split() if word not in stop_words_pt]
    stemmed_tokens = [stemmer_pt.stem(word) for word in tokens]
    return ' '.join(stemmed_tokens)

# ### ALTERAÇÃO ###: A função "w2v_vectorizer" foi completamente removida.

# ==============================================================================
# 3. DEFINIÇÃO DO EXPERIMENTO (Itens e, f, g, i, j, k)
# ==============================================================================
print("\n--- 3. Definindo o setup do experimento de classificação ---")

NUMERICAL_FEATURES = ['char_count', 'word_count', 'avg_word_length', 'punctuation_count', 'uppercase_ratio']
TEXT_FEATURE = 'text_content_anonymous'

X = df_agrupado[NUMERICAL_FEATURES + [TEXT_FEATURE]]
y = df_agrupado['is_viral']

outer_cv = KFold(n_splits=5, shuffle=True, random_state=42)
results_list = []

# --- Definindo as combinações a serem testadas ---

MODELS_TO_TEST = {
    'LogisticRegression_NoReg': LogisticRegression(penalty=None, max_iter=1000), # <--- CORRIGIDO
    'LogisticRegression_L1': LogisticRegression(penalty='l1', solver='liblinear', C=1.0),
    'LogisticRegression_L2': LogisticRegression(penalty='l2', C=1.0, max_iter=1000),
    'DecisionTree': DecisionTreeClassifier(random_state=42),
    'RandomForest': RandomForestClassifier(random_state=42)
}

SCALERS_TO_TEST = {
    'NoScaler': None,
    'Z-Score': StandardScaler(with_mean=False), # 'with_mean=False' para matrizes esparsas
    'MinMax': MinMaxScaler()
}

PREPROCESSING_TO_TEST = {
    'Sem_Preprocessamento': False,
    'Com_Preprocessamento': True
}

# ### ALTERAÇÃO ###: O dicionário de embeddings foi simplificado, removendo a opção Word2Vec.
EMBEDDINGS_TO_TEST = {
    'BOW_Unigram': ('bow', (1, 1)),
    'BOW_Bigram': ('bow', (1, 2)),
    'BOW_Trigram': ('bow', (1, 3)),
    'TFIDF_Unigram': ('tfidf', (1, 1)),
    'TFIDF_Bigram': ('tfidf', (1, 2)),
    'TFIDF_Trigram': ('tfidf', (1, 3))
}


# ==============================================================================
# 4. EXECUÇÃO DO EXPERIMENTO (LOOP PRINCIPAL)
# ==============================================================================
print("\n--- 4. Iniciando a execução do experimento com Outer K-Fold CV ---")
total_runs = len(MODELS_TO_TEST) * len(SCALERS_TO_TEST) * len(PREPROCESSING_TO_TEST) * len(EMBEDDINGS_TO_TEST)
run_count = 0

for model_name, model in MODELS_TO_TEST.items():
    for scaler_name, scaler in SCALERS_TO_TEST.items():
        for preproc_name, use_preprocessing in PREPROCESSING_TO_TEST.items():
            for embed_name, (embed_type, ngram_range) in EMBEDDINGS_TO_TEST.items():
                
                run_count += 1
                print(f"Executando ({run_count}/{total_runs}): {model_name}, {scaler_name}, {preproc_name}, {embed_name}")
                
                fold_scores = []
                
                for train_idx, test_idx in outer_cv.split(X, y):
                    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
                    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

                    X_train_text = X_train[TEXT_FEATURE]
                    X_test_text = X_test[TEXT_FEATURE]
                    if use_preprocessing:
                        X_train_text = X_train_text.apply(preprocess_text)
                        X_test_text = X_test_text.apply(preprocess_text)
                        
                    if embed_type == 'bow':
                        vectorizer = CountVectorizer(ngram_range=ngram_range)
                        X_train_text_feat = vectorizer.fit_transform(X_train_text)
                        X_test_text_feat = vectorizer.transform(X_test_text)
                    elif embed_type == 'tfidf':
                        vectorizer = TfidfVectorizer(ngram_range=ngram_range)
                        X_train_text_feat = vectorizer.fit_transform(X_train_text)
                        X_test_text_feat = vectorizer.transform(X_test_text)
                    
                    X_train_num_feat = X_train[NUMERICAL_FEATURES].values
                    X_test_num_feat = X_test[NUMERICAL_FEATURES].values
                    if scaler:
                        # Scalers como StandardScaler não funcionam bem com matrizes esparsas diretamente.
                        # Por isso, aplicamos a normalização apenas nas features numéricas antes de juntar.
                        # Esta abordagem foi implicitamente feita antes, mas agora é mais clara.
                        # MinMaxScaler funciona com matrizes esparsas, mas StandardScaler precisa de `with_mean=False`.
                        X_train_num_feat = scaler.fit_transform(X_train_num_feat)
                        X_test_num_feat = scaler.transform(X_test_num_feat)
                        
                    # ### ALTERAÇÃO ###: A lógica de combinação de features foi simplificada.
                    # Como só temos matrizes esparsas (BOW, TF-IDF), usamos hstack em todos os casos.
                    X_train_final = hstack([X_train_text_feat, X_train_num_feat]).tocsr()
                    X_test_final = hstack([X_test_text_feat, X_test_num_feat]).tocsr()

                    model.fit(X_train_final, y_train)
                    y_pred = model.predict(X_test_final)
                    
                    fold_scores.append({
                        'accuracy': accuracy_score(y_test, y_pred),
                        'precision': precision_score(y_test, y_pred, zero_division=0),
                        'recall': recall_score(y_test, y_pred, zero_division=0),
                        'f1': f1_score(y_test, y_pred, zero_division=0)
                    })

                avg_scores = pd.DataFrame(fold_scores).mean().to_dict()
                
                results_list.append({
                    'Modelo': model_name,
                    'Normalizador': scaler_name,
                    'Preprocessamento': preproc_name,
                    'Embedding_Ngram': embed_name,
                    'F1_Score_Medio': avg_scores['f1'],
                    'Precisao_Media': avg_scores['precision'],
                    'Recall_Medio': avg_scores['recall'],
                    'Acuracia_Media': avg_scores['accuracy']
                })

# ==============================================================================
# 5. ANÁLISE DOS RESULTADOS
# ==============================================================================
print("\n--- 5. Análise dos Resultados Finais ---")

results_df = pd.DataFrame(results_list)
results_df_sorted = results_df.sort_values(by='F1_Score_Medio', ascending=False).reset_index(drop=True)

print("As 15 melhores combinações de modelo e parâmetros, ordenadas por F1-Score:")
print(results_df_sorted.head(15).to_string())
```
